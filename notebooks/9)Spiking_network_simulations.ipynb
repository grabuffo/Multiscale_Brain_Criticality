# 9) Spiking-network simulations

Validates the mean-field results with the full microscopic model (Eqs. 1–2): every region is a network of active-rotator neurons.
Regions are tuned relative to the hybrid-type transition of the **spiking** model ($J=1$, $\sigma=0.499$, as in Suppl. Fig. S2), because the spiking and mean-field parameter spaces do not map one-to-one.

The simulations are long; run the commented cells from a terminal (all cores are used). Outputs go to `data/derived/`, figures to `figures/`.

## Whole-brain network of spiking neurons (Suppl. Fig. S13)

60 cortical regions × 1000 neurons, coupled through the Allen connectome.
1. Calibration: isolated populations on a grid of $a$; the critical point $a_c$ is the susceptibility peak.
2. Coupling sweep with regions at $a_c$ (critical) and $a_c + 0.005$ (subcritical), 120 s per run (~23 min per coupling value on 14 cores).
